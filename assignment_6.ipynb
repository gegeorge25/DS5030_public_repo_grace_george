```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import random
random.seed(314)
dice_roots=[]
for i in range(5000):
    dice=random.randint(1,6)
    roots_i=np.sqrt(dice)
    dice_roots.append(roots_i)
values, counts = np.unique(dice_roots,return_counts=True)
values, counts
total=np.sum(counts)
p=counts/total
p
```

```python
sns.ecdfplot(dice_roots)
plt.show()
```

```python
plt.bar(x=values,height=p,width=0.2)
plt.xlabel('Roots')
plt.ylabel('Probability')
plt.title('Probability of each Root')
```

### Sum Probabilties

```python
import random
random.seed(314)
sums=[]
for i in range(5000):
    dice1=random.randint(1,6)
    dice2=random.randint(1,6)
    sum=dice1+dice2
    sums.append(sum)
values,counts=np.unique(sums,return_counts=True)
p=counts/np.sum(counts)
p
    

```

```python
sns.ecdfplot(sums)
plt.xlabel('Sums')
plt.ylabel('Probability')
plt.show()
```

```python
plt.bar(x=values,height=p,width=0.5)
plt.xlabel('Sums')
plt.ylabel('Probability')
plt.title('Probability of each Sum')
```

- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.
