```python
import numpy as np
```

```python

rng = np.random.default_rng(100)
def dice():
    go=1. # Set go 
    sum=0. # Initialize sum as 0
    while (go==1):
        roll = rng.integers(1,7)
        sum+=roll
        if roll<6:
            go=0
    return sum
S=100000
results=[dice() for i in range(S)]
print('Expected Value', np.mean(results))
```

```python
mu=np.mean(ames['log_price'])
sig=np.std(ames['log_price'])

grid=np.linspace(9,14,100) # Based on range of log price
pdf=stats.norm.pdf(grid,mu,sig)
cdf=stats.norm.cdf(grid,mu,sig)
fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x=ames['log_price'],ax=axes[0],label='kde')
sns.lineplot(x=grid,y=pdf,color='red',ax=axes[0],label='pdf')
sns.ecdfplot(x=ames['log_price'],ax=axes[1],label='ecdf')
sns.lineplot(x=grid,y=cdf,color='red',ax=axes[1],label='cdf')
axes[0].set_title('Density')
axes[1].set_title('Distribution')


```

The PDF is more smoothed that the KDE and fits overall but has some differences.. The KDE has a highter peak around 11.5 and is more left skewed. 

The ECDF and CDF almost perfectly overlap, so the ECDF is a very good fit.

```python
x=ames['log_price']
S=1000
means=[(np.mean(x.sample(replace=True,frac=1,random_state=s))).astype('float') for s in range(S)]
sigs=[np.std(x.sample(replace=True,frac=1,random_state=s)) for s in range(S)]
sns.kdeplot(means)
plt.title("KDE of Means")
```

- Open the METABRIC data. Make a KDE of 'Overall Survival (Months)'.
- Let survival time, $T$ be distributed exponentially with parameter $\lambda$. Then its distribution is
$$
F(t) = 1 - e^{-\lambda t} = p[T\le t].
$$
- Derive the density for this distribution.
- Write out the likelihood.
- Maximize the likelihood with respect to $\lambda$. What is the MLE, $\hat{\lambda}$?
- Plot the density/distribution for your fitted model and compare it to the KDE/ECDF. Criticize the fit.
- Bootstrap the sampling density/distribution of $\hat{\lambda}$.

 Used AI for help and guidance

```python
x=metabric['Overall Survival (Months)']
```

```python
import seaborn as sns
```

```python
sns.histplot(x)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
lambda_value=len(x)/np.sum(x)
print('lambda:',lambda_value)
```

```python
grid=np.linspace(0,350,1000)
values_dist=1-np.exp(-lambda_value*grid)
```

```python
cdf=sns.lineplot(x=grid,y=values_dist)
plt.title('Exponential Distribution')
plt.xlabel('Overall Survival (Months)')
plt.ylabel('Proportion')
```

```python
values_dens=lambda_value*np.exp(-lambda_value*grid)

pdf=sns.lineplot(x=grid,y=values_dens)
plt.title('Exponential Density')
plt.xlabel('Overall Survival (Months)')
plt.ylabel('Density')
```

```python
x=metabric['Overall Survival (Months)']
mu=np.mean(x)
sig=np.std(x)
fig, axes = plt.subplots(1,2,figsize=(14,5))
sns.kdeplot(x,ax=axes[0],label='kde')
sns.lineplot(x=grid,y=values_dens,color='red',ax=axes[0],label='pdf')
sns.ecdfplot(x,ax=axes[1],label='ecdf')
sns.lineplot(x=grid,y=values_dist,color='red',ax=axes[1],label='cdf')
axes[0].set_title('Density')
axes[1].set_title('Distribution')
```

The kdes is not good fits for exponential data of overall survival months. We expect an exponential plot to start and 1 and coverage to 0, but ours does not.

The ecdf is a bit of a beter fit than the kde. It approached 0, but the cdf approaches 0 much faster.

```python
x=metabric['Overall Survival (Months)']
S=100000
lambda_values=[len(x)/np.sum(x.sample(replace=True,frac=1,random_state=s)) for s in range(S)]
sns.kdeplot(lambda_values)
plt.title('KDE of Lambda Values')
plt.xlabel('Lambda Values')
plt.show()
sns.ecdfplot(lambda_values)
plt.title('ECDF of Lambda Values')
plt.xlabel('Lambda Values')
plt.show()
```

Our bootstrapped lamba values center around 0.0078.
